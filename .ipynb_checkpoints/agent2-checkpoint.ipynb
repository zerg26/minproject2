```python
from typing import TypedDict, Optional, List, Dict, Annotated
from langgraph.graph import StateGraph, END
from langchain_core.prompts import ChatPromptTemplate
from langchain_groq import ChatGroq
from tavily import TavilyClient
import gradio as gr
import json
from dotenv import load_dotenv
import os

load_dotenv()
os.environ["GROQ_API_KEY"] = os.getenv('GROQ_API_KEY')
os.environ["TAVILY_API_KEY"] = os.getenv('TAVILY_API_KEY')
```

## Step 2: Initialize LLM and Tools

```python
llm = ChatGroq(model="llama-3.3-70b-versatile", temperature=0.3)
tavily_client = TavilyClient(api_key=os.getenv('TAVILY_API_KEY'))
```

# Agent 2: Real Estate Property Analyzer & Rater

This agent receives property listings from Agent 1, analyzes them based on user preferences, and provides ratings (1-10) for each property with detailed summaries.

```python
# Define state for Agent 2
class RealEstateAnalysisState(TypedDict):
    agent1_output: str  # Raw output from agent 1
    user_preferences: Dict  # User's preferences for weighting
    analysis: Optional[str]  # Analysis of each property
    ratings: Optional[Dict]  # Ratings for each property
    summary: Optional[str]  # Final summary
```

```python
# Agent 2 Node: Analyze and rate properties
def analyze_properties(state: RealEstateAnalysisState) -> RealEstateAnalysisState:
    """Analyzes properties from agent 1 and rates them based on user preferences"""
    
    agent1_output = state["agent1_output"]
    user_prefs = state["user_preferences"]
    
    # Create prompt for analysis
    analysis_prompt = ChatPromptTemplate.from_messages([
        ("system", """You are a real estate analyst. Analyze the properties provided and rate each one from 1-10 based on the user's preferences.
        
Consider these factors in your rating:
- Price (how well it fits the budget)
- Crime rate (safety is important)
- Commute time (convenience)
- School district (if relevant)
- Overall value and features

Weight each factor according to user preferences. Provide a detailed analysis for each property and assign a rating."""),
        ("user", """Properties from Agent 1:
{properties}

User Preferences:
{preferences}

Please analyze each property and provide:
1. A brief analysis of strengths and weaknesses
2. A rating from 1-10 (10 being the best match)
3. Key considerations for the buyer

Format your response clearly for each property.""")
    ])
    
    # Format user preferences
    pref_text = "\n".join([f"- {k}: {v}" for k, v in user_prefs.items()])
    
    # Get analysis from LLM
    chain = analysis_prompt | llm
    response = chain.invoke({
        "properties": agent1_output,
        "preferences": pref_text
    })
    
    state["analysis"] = response.content
    return state
```

```python
# Agent 2 Node: Create summary with rankings
def create_summary(state: RealEstateAnalysisState) -> RealEstateAnalysisState:
    """Creates a final summary ranking properties from best to worst"""
    
    analysis = state["analysis"]
    
    summary_prompt = ChatPromptTemplate.from_messages([
        ("system", """You are a real estate advisor. Based on the analysis provided, create a concise summary that:
1. Ranks the properties from best to worst
2. Provides a brief recommendation
3. Highlights the top choice and why

Keep it clear and actionable."""),
        ("user", """Analysis:
{analysis}

Please provide a final summary with rankings and your recommendation.""")
    ])
    
    chain = summary_prompt | llm
    response = chain.invoke({"analysis": analysis})
    
    state["summary"] = response.content
    return state
```

```python
# Build Agent 2 workflow
agent2_workflow = StateGraph(RealEstateAnalysisState)

# Add nodes
agent2_workflow.add_node("analyze", analyze_properties)
agent2_workflow.add_node("summarize", create_summary)

# Define flow
agent2_workflow.set_entry_point("analyze")
agent2_workflow.add_edge("analyze", "summarize")
agent2_workflow.add_edge("summarize", END)

# Compile
agent2_app = agent2_workflow.compile()
```

```python
# Test Agent 2 with sample data
sample_agent1_output = """Property 1: Davidson

List Price: $439,000

Neighborhood: Bradford Subdivision

Crime Rate (Area): Very Low (A- rating from AreaVibes)

Avg. Commute to Uptown: 25-35 minutes

School District (GreatSchools): 9/10

Notes: 4 beds, 2.5 baths. Near Lake Norman and a walkable historic downtown. High HOA fees.

Property 2: South Charlotte (Piper Glen)

List Price: $449,900

Neighborhood: Piper Glen Estates

Crime Rate (Area): Low (B+ rating from AreaVibes)

Avg. Commute to Uptown: 20-30 minutes

School District (GreatSchools): 8/10

Notes: 3 beds, 3 baths. Backs onto a golf course. Mature community with established landscaping.

Property 3: East Charlotte (Oakhurst)

List Price: $425,000

Neighborhood: Oakhurst

Crime Rate (Area): Moderate (C rating from AreaVibes)

Avg. Commute to Uptown: 15-20 minutes

School District (GreatSchools): 6/10

Notes: 4 beds, 2 baths. Recently renovated. Appreciating area but still has some property crime concerns."""

sample_user_preferences = {
    "budget_priority": "High - want best value under $450k",
    "safety_priority": "Very High - low crime is essential",
    "commute_priority": "Medium - prefer under 30 min",
    "schools_priority": "High - good schools important for resale",
    "other_notes": "Willing to pay more for safety and schools"
}

# Run Agent 2
initial_state = {
    "agent1_output": sample_agent1_output,
    "user_preferences": sample_user_preferences,
    "analysis": None,
    "ratings": None,
    "summary": None
}

result = agent2_app.invoke(initial_state)
```
